**Resumen Sesion 12**

Los árboles de decisión son modelos que toman datos y los van separando mediante preguntas hasta obtener una respuesta. Su estructura empieza en una raíz, continúa con evaluaciones y desciende por diferentes ramas según cada respuesta.

El algoritmo CART sirve para clasificación y regresión, buscando realizar buenas divisiones de los datos. La entropía indica qué tan desordenada está la información. Como los árboles pueden crecer demasiado y causar sobreajuste, se utiliza la regularización para limitar su crecimiento.

Los árboles de regresión buscan encontrar grupos de datos más parecidos. Una desventaja de los árboles es que sus decisiones son bastante rígidas, por lo que combinar varios árboles y sus resultados puede mejorar las predicciones.
