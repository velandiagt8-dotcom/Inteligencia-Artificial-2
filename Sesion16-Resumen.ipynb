**Resumen: Redes neuronales**

Las redes neuronales artificiales están inspiradas en cómo funciona el cerebro. Reciben información, la procesan y van aprendiendo qué datos son importantes. El perceptrón fue uno de los primeros modelos, pero era muy limitado. Luego aparecieron las redes con varias capas (MLP), que pueden aprender cosas más complicadas.

Para aprender, usan backpropagation, que básicamente revisa los errores y ajusta la red para mejorar. Se pueden usar para reconocer imágenes, clasificar objetos y encontrar patrones. Aunque se inspiran en el cerebro, no funcionan igual, ya que las redes artificiales trabajan principalmente con matemáticas y datos.
