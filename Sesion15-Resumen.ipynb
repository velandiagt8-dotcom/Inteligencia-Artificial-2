**Resumen: Aprendizaje no supervisado**

El aprendizaje no supervisado permite que una máquina analice datos sin que previamente le digamos qué significa cada uno. Su objetivo es encontrar similitudes, diferencias y patrones por cuenta propia. Por ejemplo, puede separar datos en grupos usando K-means, reconocer diferentes partes de una imagen o encontrar información que se sale de lo normal. También existen métodos como DBSCAN y GMM, que ayudan a trabajar con grupos de diferentes formas y a detectar datos extraños. En general, sirve para organizar información, descubrir patrones y encontrar datos importantes sin necesidad de tener todo clasificado desde el principio.
