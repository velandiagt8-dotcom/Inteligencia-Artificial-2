**RESUMEN SESION 4**

La Ruta de Ciencia de Datos es un proceso que permite utilizar los datos para comprender problemas, encontrar patrones y tomar mejores decisiones. Primero se define el problema y el objetivo que se quiere alcanzar. Luego se recopilan datos de diferentes fuentes, como bases de datos, aplicaciones, encuestas o páginas web.

Después, los datos se limpian y preparan, corrigiendo errores, eliminando duplicados y organizando la información. Luego se realiza un análisis para encontrar tendencias, relaciones y características importantes. Con estos datos se entrena un modelo de Machine Learning, que puede utilizar aprendizaje supervisado, no supervisado o por refuerzo, dependiendo del problema.

Finalmente, el modelo se evalúa con datos nuevos para comprobar sus resultados y, si es necesario, se realizan ajustes. Cuando funciona correctamente, puede aplicarse en situaciones reales, aunque debe revisarse y actualizarse porque los datos pueden cambiar con el tiempo. En resumen, el Machine Learning es parte de una ruta que va desde la recolección y preparación de datos hasta la creación de modelos capaces de realizar predicciones y encontrar patrones.
