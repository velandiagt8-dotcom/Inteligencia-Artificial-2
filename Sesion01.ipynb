# 🤖 Machine Learning e Inteligencia Artificial

## 🧠 Conceptos principales

### Machine Learning
Es una rama de la Inteligencia Artificial que permite que una computadora aprenda patrones a partir de datos y haga predicciones o tome decisiones.

### Modelar patrones no lineales
Consiste en encontrar relaciones complejas entre los datos que no pueden representarse simplemente con una línea recta.

### Ecosistema ecológico
Es un conjunto de elementos que interactúan dentro de un entorno. En Machine Learning puede utilizarse como ejemplo para analizar relaciones entre diferentes variables.

### Fuga de datos (Data Leakage)
Ocurre cuando información que no debería estar disponible durante el entrenamiento del modelo termina influyendo en él, haciendo que parezca más preciso de lo que realmente es.

---

## 🔄 Flujo de procesamiento de datos

**Línea → Imputación → Escalamiento → Reducción → Clasificador**

- **Imputación:** completar los valores que faltan.
- **Escalamiento:** ajustar los valores de las variables a una escala común.
- **Reducción:** disminuir la cantidad de variables o dimensiones.
- **Clasificador:** utilizar un modelo para asignar los datos a diferentes categorías.

---

## 🎯 Modelos supervisados

Son modelos que aprenden utilizando datos que ya tienen una respuesta conocida.

**Ejemplo:** utilizar información de pacientes para predecir a qué categoría pertenece un nuevo paciente.

---

## 🌀 Truco de Kernel

Es una técnica utilizada principalmente en modelos como **SVM** para trabajar con relaciones no lineales. Permite transformar los datos a un espacio donde sea más fácil separarlos.

---

## 🎒 Bagging

Consiste en crear varios modelos y combinar sus resultados para obtener una predicción más estable y reducir la variabilidad.

---

## 🚀 Boosting

Construye modelos de manera secuencial, haciendo que cada nuevo modelo intente corregir los errores cometidos por los modelos anteriores.

---

## 🔍 Inteligencia Artificial no supervisada

Trabaja con datos que **no tienen una respuesta previamente definida** y busca patrones, grupos o estructuras dentro de la información.

---

## 📊 Dimensionalidad y varianza

### Dimensionalidad
Es la cantidad de variables o características que tiene un conjunto de datos.

La reducción de dimensionalidad puede hacer que los modelos sean más rápidos y fáciles de interpretar.

### Varianza
Mide qué tanto cambian los resultados de un modelo cuando cambian los datos utilizados para entrenarlo.

---

## 📈 Evaluación de modelos

Es el proceso de comprobar qué tan bien funciona un modelo utilizando datos de prueba y diferentes métricas.

### Métricas

Algunas métricas utilizadas son:

- **Accuracy:** exactitud general.
- **Precision:** qué tan precisas son las predicciones positivas.
- **Recall:** qué proporción de los casos positivos logra detectar.
- **F1-Score:** equilibrio entre precision y recall.
- **ROC-AUC:** mide la capacidad del modelo para distinguir entre clases.

---

## ⚖️ Sesgo y varianza

- **Sesgo (Bias):** aparece cuando el modelo es demasiado simple y no logra captar correctamente los patrones.
- **Varianza (Variance):** aparece cuando el modelo es demasiado sensible a los datos de entrenamiento.

### ⚖️ Equilibrio

Se busca encontrar un equilibrio entre **sesgo y varianza** para evitar:

- **Underfitting:** el modelo aprende muy poco.
- **Overfitting:** el modelo aprende demasiado los datos de entrenamiento y funciona mal con datos nuevos.

---

## 🔎 GridSearchCV

Es una técnica que prueba sistemáticamente diferentes combinaciones de parámetros de un modelo para encontrar la configuración que obtiene mejores resultados.

---

## 🎲 RandomizedSearchCV

Es similar a GridSearchCV, pero selecciona aleatoriamente diferentes combinaciones de parámetros.

Puede ser más rápido cuando existen muchas combinaciones posibles.

---

## 🏭 IA predictiva en la Industria 5.0

La Inteligencia Artificial predictiva permite analizar grandes cantidades de datos para **anticipar situaciones, detectar problemas y apoyar la toma de decisiones**.

En la **Industria 5.0**, busca combinar la tecnología y la automatización con la participación humana, haciendo los procesos más inteligentes, eficientes y adaptables.

# 💡 En resumen

Machine Learning permite que las computadoras aprendan de grandes cantidades de información para encontrar patrones y realizar predicciones. Para lograrlo, primero se preparan los datos, se pueden reducir sus dimensiones y posteriormente se utilizan modelos supervisados o no supervisados. También se estudian técnicas como **Kernel, Bagging y Boosting**, además de conceptos como **sesgo, varianza, fuga de datos y evaluación de modelos**. Finalmente, herramientas como **GridSearchCV y RandomizedSearchCV** ayudan a encontrar mejores configuraciones para los modelos, permitiendo aplicar la IA predictiva a áreas como la **Industria 5.0**.
