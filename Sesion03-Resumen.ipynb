**RESUMEN SESION 03**

El KDD es una forma de pasar de tener muchos datos desordenados a obtener algo que realmente nos sirva. Tener miles de datos no significa que ya tengamos conocimiento, porque primero hay que seleccionarlos, corregir errores, organizarlos y analizarlos. El proceso se hizo más conocido con los trabajos de Fayyad en 1996 y tiene cinco pasos: seleccionar los datos, limpiarlos, organizarlos, buscar patrones y revisar los resultados. Así podemos encontrar cosas útiles, como saber qué productos prefieren las personas o hacer una recomendación. Con el Big Data, este proceso es aún más importante porque se manejan cantidades enormes de información. En resumen, los datos son como materia prima y el KDD ayuda a convertirlos en conocimiento útil para tomar mejores decisiones, siempre teniendo cuidado con la información privada o sensible
