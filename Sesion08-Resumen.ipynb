**RESUMEN: Machine Learning de extremo a extremo**

El Machine Learning de principio a fin busca tomar datos que están desordenados y convertirlos en información útil para tomar buenas decisiones. Primero se reúnen y organizan los datos, luego se limpian y preparan para que el sistema pueda entenderlos. Después se crea y entrena un modelo, se revisa que no cometa errores al trabajar con datos nuevos y se comprueba si realmente funciona en situaciones reales. Finalmente, el modelo se debe actualizar con nuevos datos para que siga dando buenos resultados.
