**ALGORITMOS**

**1. Árbol de decisión (Decision Tree):** Es un algoritmo que sirve para tomar decisiones a partir de diferentes condiciones. Va haciendo preguntas y, dependiendo de las respuestas, llega a un resultado. Se usa, por ejemplo, para decidir si una persona puede recibir un crédito, clasificar clientes o predecir una situación.

**2. K-means:** Es un algoritmo que sirve para agrupar datos que tienen características similares. El sistema encuentra grupos dentro de los datos sin que estos tengan que estar clasificados previamente. Se puede usar, por ejemplo, para agrupar clientes según sus gustos, compras o comportamientos.

**3. K-nearest neighbors (KNN):** Este algoritmo clasifica un dato nuevo comparándolo con otros datos que son similares o que están cerca de él. Se utiliza en casos como la clasificación de imágenes, identificación de categorías o para determinar a qué grupo pertenece un nuevo cliente.

**4. Random Forest:** Es un algoritmo que combina varios árboles de decisión para obtener resultados más confiables. Se utiliza principalmente para clasificar datos y hacer predicciones, por ejemplo, para detectar si una transacción puede ser sospechosa o para predecir determinados comportamientos.

**5. Naive Bayes:** Es un algoritmo que calcula qué tan probable es que un dato pertenezca a una determinada categoría. Se utiliza mucho para trabajar con textos, por ejemplo, para identificar si un correo electrónico es spam, clasificar mensajes o analizar diferentes tipos de textos.

**6. Support Vector Machine (SVM):** Este algoritmo busca separar diferentes grupos de datos utilizando un límite que permita distinguirlos de la mejor manera posible. Se puede utilizar para clasificar imágenes, textos y diferentes tipos de información en categorías.

**7. AdaBoost:** Es un algoritmo que combina varios modelos sencillos para conseguir un resultado más preciso. Cada nuevo modelo intenta mejorar los errores cometidos anteriormente. Se utiliza principalmente en problemas donde se necesita mejorar la clasificación de diferentes datos.

**8. Q-learning:** Es un algoritmo que permite que un sistema aprenda mediante prueba y error. El sistema recibe recompensas cuando realiza buenas acciones y aprende cuáles son las mejores decisiones. Se utiliza en videojuegos, robots y sistemas que necesitan aprender qué hacer dependiendo de la situación.

**9. Red neuronal convolucional (CNN):** Es un tipo de algoritmo especialmente diseñado para trabajar con imágenes y reconocer diferentes características dentro de ellas. Se utiliza en reconocimiento facial, identificación de objetos, análisis de imágenes y sistemas de visión artificial.

**10. DBSCAN:** Es un algoritmo que agrupa datos que se encuentran cerca y tienen características similares. También permite encontrar datos que son diferentes o que no pertenecen a ningún grupo. Se utiliza, por ejemplo, para encontrar grupos de clientes y detectar comportamientos o datos poco comunes.
