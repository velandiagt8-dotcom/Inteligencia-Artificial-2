**Resumen: Entrenamiento de redes profundas**

El entrenamiento de redes profundas puede tener problemas porque los gradientes pueden desaparecer o explotar al pasar por muchas capas. Para solucionarlo se usan técnicas como Glorot/Xavier para inicializar bien los pesos, ReLU y sus variantes para evitar neuronas muertas, y la normalización para estabilizar los datos.

También se utiliza gradient clipping para controlar gradientes demasiado grandes. El aprendizaje transferido aprovecha redes ya entrenadas y el dropout ayuda a evitar que el modelo dependa demasiado de unas pocas neuronas. Todo esto permite entrenar redes más profundas de forma más estable y efectiva.
