RESUMEN

El Machine Learning o aprendizaje automático permite que las máquinas aprendan a partir de datos para reconocer patrones, hacer predicciones y tomar decisiones sin tener que programar cada situación. Dentro de este tema se encuentran la minería de datos, que busca información útil entre muchos datos, y diferentes tipos de aprendizaje, como el supervisado, no supervisado y por refuerzo, donde el sistema puede aprender mediante ejemplos, encontrar patrones por sí mismo o recibir recompensas y penalizaciones. También existen formas de trabajar con los datos, como el aprendizaje por lotes, que utiliza grupos de datos, y el aprendizaje en línea, que aprende mientras recibe información nueva. Algunos modelos comparan situaciones con ejemplos anteriores (basado en instancias) y otros crean una forma de representar los datos para hacer predicciones (basado en modelos). Uno de los objetivos principales es la generalización, es decir, que el modelo pueda funcionar bien con datos nuevos. Sin embargo, existen problemas como el sobreajuste, cuando el modelo memoriza los datos y hasta sus errores, y el infrajuste, cuando aprende muy poco. Además, se debe tener en cuenta que “si entra basura, sale basura”, porque datos incorrectos o de mala calidad producen malos resultados. Por eso los modelos se evalúan mediante entrenamiento, validación y prueba final, comprobando que realmente aprendan y funcionen correctamente. En los modelos no supervisados también existe el riesgo de encontrar patrones o grupos que realmente no sean útiles o que lleven a conclusiones equivocadas.
